````python
# Family Dynamics: A Directed Graph of Relationships and Alliances
# Developed using python by Heider Jeffer

import networkx as nx
import matplotlib.pyplot as plt

# Create a directed graph
G = nx.DiGraph()

# Define family members (for example purposes)
family_members = [
    "Estifo (Father)", "Marilyn Monroe (Mother)", "Hiyam (Eldest Daughter)", 
    "Antoine (Son, Officer)", "Edward (Son, Schemer)", "Emile (Son, Betrayer)",
    "Ivonne (Daughter, Gullible)", "Clara (Daughter, Gullible+)", 
    "Badria (Antoine's Wife)", "Mirna (Antoine's Daughter)", "Lina (Edward's Wife)"
]

# Add nodes (family members)
G.add_nodes_from(family_members)

# Define relationships (edges) between family members
relationships = [
    ("Estifo (Father)", "Hiyam (Eldest Daughter)"), 
    ("Estifo (Father)", "Clara (Daughter, Gullible+)"), 
    ("Estifo (Father)", "Ivonne (Daughter, Gullible)"), 
    ("Estifo (Father)", "Antoine (Son, Officer)"),
    ("Estifo (Father)", "Edward (Son, Schemer)"), 
    ("Estifo (Father)", "Emile (Son, Betrayer)"),
    ("Marilyn Monroe (Mother)", "Hiyam (Eldest Daughter)"), 
    ("Marilyn Monroe (Mother)", "Clara (Daughter, Gullible+)"), 
    ("Marilyn Monroe (Mother)", "Ivonne (Daughter, Gullible)"), 
    ("Marilyn Monroe (Mother)", "Antoine (Son, Officer)"),
    ("Marilyn Monroe (Mother)", "Edward (Son, Schemer)"), 
    ("Marilyn Monroe (Mother)", "Emile (Son, Betrayer)"),
    ("Antoine (Son, Officer)", "Badria (Antoine's Wife)"),
    ("Antoine (Son, Officer)", "Mirna (Antoine's Daughter)"),
    ("Edward (Son, Schemer)", "Lina (Edward's Wife)")
]

# Add edges (relationships)
G.add_edges_from(relationships)

# Improve the visualization of the family relationship graph
plt.figure(figsize=(14, 10))
pos = nx.spring_layout(G, seed=42)  # Consistent positioning

# Define node colors for clarity
node_colors = {
    "Estifo (Father)": "lightcoral",
    "Marilyn Monroe (Mother)": "lightpink",
    "Hiyam (Eldest Daughter)": "gold",
    "Antoine (Son, Officer)": "lightblue",
    "Edward (Son, Schemer)": "orange",
    "Emile (Son, Betrayer)": "purple",
    "Ivonne (Daughter, Gullible)": "lightgreen",
    "Clara (Daughter, Gullible+)": "lightgreen",
    "Badria (Antoine's Wife)": "red",
    "Mirna (Antoine's Daughter)": "darkred",
    "Lina (Edward's Wife)": "gray"
}

# Apply colors to nodes
node_color_list = [node_colors.get(node, "white") for node in G.nodes]

# Draw graph elements
nx.draw_networkx_nodes(G, pos, node_color=node_color_list, node_size=2500, edgecolors="black")
nx.draw_networkx_edges(G, pos, arrowstyle="->", arrowsize=15, edge_color="black", width=1.5)
nx.draw_networkx_labels(G, pos, font_size=10, font_family="Arial", font_weight="bold")

# Draw edge labels with relationships (if needed)
edge_labels = {(u, v): d.get("label", "") for u, v, d in G.edges(data=True)}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=9, font_family="Arial", font_color="darkblue")

# Title
plt.title("Family Relationship & Behavior Analysis", fontsize=14, fontweight="bold")
plt.show()
````

### **Explanation of the Code: Family Relationship Chart**
*By Heider Jeffer*

This Python script **visualizes a family tree** using the `networkx` and `matplotlib` libraries. It represents **family members as nodes** and **their relationships as directed edges** in a graph. Below is a breakdown of the code:

---

### **1. Import Necessary Libraries**
```python
import networkx as nx
import matplotlib.pyplot as plt
```
- `networkx` is used for creating and managing graphs.
- `matplotlib.pyplot` is used for drawing the graph.

---

### **2. Create a Directed Graph**
```python
G = nx.DiGraph()
```
- A **directed graph (DiGraph)** is created, meaning the edges have direction (e.g., parent to child).

---

### **3. Define Family Members**
```python
family_members = [
    "Astivo", "Marilyn Monroe", "Hiam", "Clara", "Ivonne", "Antoine", "Edward", "Emil",
    "Badria", "Mirna", "Lina"
]
```
- A list of **family members** is defined, each name representing a node in the graph.

---

### **4. Add Nodes to the Graph**
```python
G.add_nodes_from(family_members)
```
- Adds all family members as **nodes** in the graph.

---

### **5. Define Relationships (Edges)**
```python
relationships = [
    ("Astivo", "Hiam"), ("Astivo", "Clara"), ("Astivo", "Ivonne"), ("Astivo", "Antoine"), ("Astivo", "Edward"), ("Astivo", "Emil"),
    ("Marilyn Monroe", "Hiam"), ("Marilyn Monroe", "Clara"), ("Marilyn Monroe", "Ivonne"), ("Marilyn Monroe", "Antoine"), ("Marilyn Monroe", "Edward"), ("Marilyn Monroe", "Emil"),
    ("Antoine", "Badria"), ("Antoine", "Mirna"),
    ("Edward", "Lina")
]
```
- This list contains **directed edges** representing relationships:
  - `"Astivo"` (Father) is connected to all his children.
  - `"Marilyn Monroe"` (Mother) is also connected to all her children.
  - `"Antoine"` (Son) is connected to his **wife ("Badria")** and **daughter ("Mirna")**.
  - `"Edward"` (Son) is connected to his **wife ("Lina")**.

---

### **6. Add Relationships to the Graph**
```python
G.add_edges_from(relationships)
```
- Adds all **edges** (relationships) to the graph.

---

### **7. Draw the Graph**
```python
plt.figure(figsize=(10, 6))  # Set figure size
pos = nx.spring_layout(G, seed=42)  # Generate positions for nodes
nx.draw(G, pos, with_labels=True, node_color='lightblue', edge_color='gray', 
        node_size=3000, font_size=10, font_weight='bold')  # Draw nodes and edges
plt.title("Family Relationship Chart")  # Add title
plt.show()  # Display the graph
```
- `plt.figure(figsize=(10, 6))`: Sets the **size** of the graph.
- `pos = nx.spring_layout(G, seed=42)`: Generates **positions** for the nodes using the **spring layout** algorithm.
- `nx.draw(...)`: Draws the **nodes, edges, and labels** with the following settings:
  - `node_color='lightblue'`: Makes the nodes **light blue**.
  - `edge_color='gray'`: Makes the edges **gray**.
  - `node_size=3000`: Adjusts node size.
  - `font_size=10, font_weight='bold'`: Improves text readability.
- `plt.title(...)`: Adds a **title** to the graph.
- `plt.show()`: Displays the graph.

---

### **📌 What This Code Does:**
✅ **Creates a family tree-like structure** using a directed graph.  
✅ **Shows parental relationships** from **Astivo & Marilyn Monroe** to their children.  
✅ **Visualizes connections** between Antoine and his wife/daughter, and Edward and his wife.  
✅ **Uses color and layout styling** for a clear and readable graph.
